# RankLab: Open Bandit Dataset OPE

This is isolated from KuaiRand-Pure. It uses OBD's recorded propensities and separate random-policy feedback; it never invents KuaiRand propensities.

```python
from pathlib import Path
import subprocess

repo = Path('/kaggle/working/rank-lab')
subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/kushc2004/rank-lab.git', str(repo)], check=True)
subprocess.run(['pip', 'install', '--quiet', '-e', '.[full,ope]'], cwd=repo, check=True)
subprocess.run([
    'python', 'scripts/run_ope_benchmark.py', '--campaign', 'all', '--behavior-policy', 'bts',
    '--output', 'outputs/metrics/ope.json',
], cwd=repo, check=True)
subprocess.run(['tar', '-czf', '/kaggle/working/ranklab_open_bandit_ope_outputs.tar.gz', '-C', str(repo), 'outputs'], check=True)
print('Saved /kaggle/working/ranklab_open_bandit_ope_outputs.tar.gz', flush=True)
```
